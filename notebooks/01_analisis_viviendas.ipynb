```python
# Vamos a trabajar con las casas de California
import pandas as pd
from sklearn.datasets import fetch_california_housing
```

```python
# Cargamos el dataset, as_frame=True nos devuelve un DataFrame de pandas, no un arreglo de Numpy,
# metodo frame incluye la columna objetivo (target) en el DataFrame, mientras que data solo 
# incluye las características (features).
df = fetch_california_housing(as_frame=True).frame
pd.set_option('display.max_columns', None)  # Mostrar todas las columnas, None significa sin límite
pd.set_option('expand_frame_repr', False) # Evita que el DataFrame se divida en varias líneas al imprimirlo
print(df.head(10))
```

El dataset de California Housing recopiló datos del censo de 1990 en California. Un detalle muy importante es que cada fila de la tabla no representa una sola casa, sino un "bloque de casas" (un grupo vecinal o manzana que promedia unas 1,425 personas).
MedInc (Median Income): El ingreso medio de las familias que viven en ese bloque (está medido en decenas de miles de dólares, por ejemplo: 3.5 significa $35,000 dólares al año). Es la variable más importante para predecir el precio.
HouseAge (Housing Median Age): La edad mediana de las viviendas en ese bloque (cuántos años tienen de construidas).
AveRooms (Average Rooms): El número promedio de habitaciones (cuartos totales) por hogar en ese bloque.
AveBedrms (Average Bedrooms): El número promedio de dormitorios (recámaras para dormir) por hogar en ese bloque.
Population (Population): El número total de personas que residen en ese bloque de viviendas.
AveOccup (Average Occupancy): El número promedio de miembros o personas que viven por hogar en ese bloque.
Latitude (Latitude): La coordenada de latitud geográfica del centro del bloque (hacia el norte/sur).
Longitude (Longitude): La coordenada de longitud geográfica del centro del bloque (hacia el este/oeste).
MedHouseVal (Median House Value): El precio medio de las casas en ese bloque. Esta es la variable objetivo (y), está medida en cientos de miles de dólares (ejemplo: 2.0 significa $200,000 dólares)

```python
# Vamos a sacar los outliers de la columna AveRooms
df = df[df['AveRooms'] <= 10]  # Filtramos los valores de AveRooms menores o iguales a 10
print(df.describe())  # Mostramos las estadísticas descriptivas después de eliminar los outliers
print(df.info())  # Mostramos la información del DataFrame después de eliminar los outliers
```
