```python
# Vamos a hacer un analisis de colinealidad de los datos de California
import pandas as pd
from sklearn.datasets import fetch_california_housing

df = fetch_california_housing(as_frame=True).frame
```

```python
df = df[df['AveRooms'] <= 10]
df = df[df['AveOccup'] <= 10]
df = df[df['Population'] <= 6000]
df = df[df['MedInc'] <= 11]
```

```python
# Graficamos un mapa de calor de la matriz de colinealidad
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Matriz de Colinealidad')
plt.show()
```

```python
# grafico de distribuciones de la Matriz de Colinealidad
plt.figure(figsize=(10, 8))
sns.pairplot(df)
```
