```python
# Vamos a hacer un analisis de colinealidad de los datos de California
import pandas as pd
from sklearn.datasets import fetch_california_housing

df = fetch_california_housing(as_frame=True).frame
```

```python
# Graficamos un mapa de calor de la matriz de colinealidad
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Matriz de Colinealidad')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))
# Dibujamos los puntos usando longitud en X y latitud en Y
# El color (c) cambia según el precio de la casa y el tamaño (s) según la población
plt.scatter(df['Longitude'], df['Latitude'], alpha=0.4,
            s=df['Population']/100, label='Población',
            c=df['MedHouseVal'], cmap='jet')

plt.colorbar(label='Precio Medio de la Casa (MedHouseVal)')
plt.xlabel('Longitud')
plt.ylabel('Latitud')
plt.title('Mapa Geográfico de Precios en California')
plt.legend()
plt.show()
```
